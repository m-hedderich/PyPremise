# 🧩 PyPremise Logging Tutorial

This short tutorial shows how to use the built-in logging functionality in **PyPremise**.
Logging helps you understand what happens inside the library — such as which Premise engine is used, where temporary files are written, and when the C++ executable starts running.

By adjusting the log level, you can make PyPremise more or less verbose depending on your needs.

## 1️⃣ Import and setup logging

The default log level in Python is `WARNING`, which means you will **not** see messages logged with `INFO` or `DEBUG`.
You can change this behavior using `logging.basicConfig(level=logging.INFO)` or `level=logging.DEBUG`.

````python
import logging

# Show all INFO and above messages
logging.basicConfig(level=logging.INFO, format="%(levelname)s: %(message)s")
````

## 2️⃣ Run a simple PyPremise example

Let's use the default synthetic example shipped with the package. If you have your own dataset, you can replace this part later.

````python
from pypremise import Premise, data_loaders

premise_instances,  _, _ = data_loaders.get_dummy_data()

premise = Premise()

# Run the Premise pattern search (this will trigger various log messages)
results = premise.find_patterns(premise_instances)

for r in results:
    print(r)
````

You should now see several log lines like:
```
INFO: Starting Premise. This might take a while.
INFO: Premise command: Premise_Windows.exe ...
INFO: Result file size: 123 bytes
```
If any temporary file cannot be removed, you'll see a warning like:
```
WARNING: Could not remove /tmp/tmpabcd1234: [WinError 32] The process cannot access the file...
```

## 3️⃣ Change verbosity

Try switching between different log levels to control the verbosity:

````python
logging.getLogger().setLevel(logging.WARNING)
print("Now only warnings and errors will be shown.")

# Run again with fewer logs
results = premise.find_patterns(premise_instances)
````

## 4️⃣ Available log levels

| Level | Meaning |
|-------|----------|
| `DEBUG` | Detailed internal debug information (developer-level). |
| `INFO` | General information about the process (recommended). |
| `WARNING` | Something unexpected happened, but execution continues. |
| `ERROR` | A serious problem occurred (e.g. Premise executable failed). |
| `CRITICAL` | The program cannot continue. |

You can use:
```python
logging.getLogger().setLevel(logging.DEBUG)
```
to see **even more detailed** internal messages, for example when PyPremise automatically detects which Premise engine (Linux/Mac/Windows) to use.

## ✅ Summary

- Use `logging.basicConfig(level=logging.INFO)` at the start of your script to view what PyPremise is doing internally.
- Switch to `DEBUG` for troubleshooting or development.
- Log messages are printed consistently across modules (`core.py`, `io.py`, etc.), so you can easily trace what happens under the hood.

This makes PyPremise more transparent and helps prevent silent issues such as undeleted temporary files or misconfigured executables.
